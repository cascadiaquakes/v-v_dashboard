```python
import pandas as pd
# import matplotlib.pyplot as plt
from plotly.subplots import make_subplots
import plotly.graph_objects as go
```

```python
# Define a function to convert seconds to years, days, hours, and seconds
def convert_seconds_to_time(seconds):
    years = seconds / (365.25 * 24 * 3600)
    days = (seconds / (24 * 3600))
    hours = seconds / 3600
    seconds = seconds
    return years, days, hours, seconds
```

```python
df = pd.read_parquet("../../../Downloads/BL13M_instant_gauge00001.parquet")
df['years'], df['days'], df['hours'], df['seconds'] = convert_seconds_to_time(df['t'])
df
```

```python
fig = make_subplots(rows=1, cols=2, shared_xaxes=True,
                    subplot_titles=['depth', 'surf'],
                   vertical_spacing=0.08, horizontal_spacing=0.1)
# Add traces for each variable
fig.add_trace(go.Scatter(x=df['t'], y=df['depth'], mode='lines', name='Depth'),   row=1, col=1)
fig.add_trace(go.Scatter(x=df['t'], y=df['surf'], mode='lines', name='surface'), row=1, col=2)

# Update layout
fig.update_layout(title='Variables over Time', showlegend=False, height=1000)
fig.update_xaxes(matches='x')
# Show the figure
fig
```
